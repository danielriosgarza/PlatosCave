## Data files

A notebook's data files lie in a folder with the notebook's name, beside it: `notebooks/demo/` for `notebooks/demo.ipynb`. When the notebook runs, a file's path is its path inside that folder, so `notebooks/demo/data/heights.csv` is opened as `data/heights.csv`.

```python
import csv
from statistics import mean, median

with open('data/heights.csv') as file:
    heights = [float(row['height_cm']) for row in csv.DictReader(file)]

print(f'{len(heights)} plants')
print(f'mean   {mean(heights):.1f} cm')
print(f'median {median(heights):.1f} cm')
```

One plant is much taller than the others. It pulls the mean up and leaves the median where it was.

```python
import matplotlib.pyplot as plt

plt.rcParams.update({'font.family': 'DejaVu Sans', 'font.size': 11})
figure, axes = plt.subplots(figsize=(6, 2.6), dpi=100)
axes.hist(heights, bins=range(10, 34), color='white', edgecolor='black')
axes.axvline(mean(heights), color='black', linestyle='--', label='mean')
axes.axvline(median(heights), color='black', linestyle=':', label='median')
axes.set_xlabel('Height (cm)')
axes.set_ylabel('Plants')
axes.legend(frameon=False)
axes.spines[['top', 'right']].set_visible(False)
figure.tight_layout()
plt.show()
```
